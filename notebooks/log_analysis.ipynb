```python
import re
import pandas as pd

LINE = re.compile(
    r"(?P<month>\w{3})\s+(?P<day>\d+)\s+(?P<time>[\d:]+)\s+"
    r"\S+\s+sshd\[\d+\]:\s+(?P<result>Accepted|Failed)\s+password\s+"
    r"for\s+(?:invalid user\s+)?(?P<user>\S+)\s+from\s+(?P<ip>\d+\.\d+\.\d+\.\d+)"
)

rows = []
with open('../data/auth.log') as f:
    for line in f:
        m = LINE.search(line)
        if m:
            rows.append(m.groupdict())
df = pd.DataFrame(rows)
df.head()
```

```python
# Failed vs Accepted
df['result'].value_counts()
```

```python
# Failed logins per IP
failed = df[df['result'] == 'Failed']
failed.groupby('ip').size().sort_values(ascending=False)
```
